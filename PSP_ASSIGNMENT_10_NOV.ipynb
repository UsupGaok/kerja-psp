```python
import time
import random

day = 1
health = 100
inventory = []
energy = 150
food_supply = 0
shelter = False
searching_for_help_attempt = 0

#Tuple (fixed locations data)
LOCATIONS=("Waterfall","Dark cave","Dense forest")

#For loading the print and make the output appeared more slowly
def blink(name, duration):
  for i in range(len(name)):
    time.sleep(duration)
    print(name[i], end ="" , flush = True)
  print()
  return "" # to eliminate output of None when using this in input func

#Fight bear scenario
def fight_bear():
    global health, food_supply, energy
    if 'Knife' in inventory :
        blink("""
U DEFEATED THE BEAR
---> HEALTH -25
""", 0.035)
        health -= 25
        food_supply += 2
        energy -= 20

    elif not "Knife" in inventory :
        blink("""
U DEFEATED THE BEAR
---> HEALTH -50
""", 0.035)
        health -= 50
        food_supply += 2
        energy -= 40

#Random event generator
def random_event():
  global health,food_supply
  event_chance=random.randint(1,3)
  if event_chance == 1:
    blink("\n🎲 random event: You found a wild fruit tree!🍇 (+1 food supply)",0.035)
    food_supply += 1
  elif event_chance == 2:
    blink("\n🎲 random event: You stepped on a sharp thorn! 🦶🏻 (-10 Health)",0.035)
    health -= 10


#Searching for food
def search_food() :
    global food_supply, energy
    blink('SEARCHING FOR FOOD....FOOD FOUND', 0.035)
    blink("""
---> FOOD SUPPLY +1
---> ENERGY -20
""", 0.035)
    food_supply += 1
    energy -= 20
    random_event()

#For building a shelter
def built_shelter() :
    global energy
    if "Rope" in inventory :
        blink("BUILDING A SHELTER WITH ROPE....DONEEE!⛺✅✅✅", 0.035)
        blink("---> ENERGY -30", 0.035)
        energy -= 30

    elif not "Rope" in inventory :
        blink("BUILDING A SHELTER WITH BARE HAND....DONEEE!⛺✅✅✅", 0.035)
        blink("---> ENERGY -50", 0.035)
        energy -= 50

def status_show():
  global shelter
  shelter_valid = "✅" if shelter else "❌"
  blink(f"""
============================================
--- DAY {day} ---
Player name : {player_name}
Health      : {health}
Energy      : {energy}
Food supply : {food_supply}
Inventory   : {inventory}
Shelter     : {shelter_valid}
=============================================
""", 0.02)
#============================================
#START THE GAME & SELECT ITEM
#============================================
print("****************************************")
print("  PYTHON QUEST : SURVIVE IN THE ISLAND   ")
print("****************************************")

#Name selection
player_name = input(blink('Enter your name : ', 0.035)).strip()



#Item selection

while not inventory :
    item_select = input(blink("""
ITEM SELECTION

A ----> [Lighter, Knife, Rope]
B ----> [First aid kit, Knife, Lighter]
C ----> [Lighter, First aid kit, Rope]

Please select your item(A/B/C) : """, 0.035))

    if item_select.upper().strip() == "A" :
        inventory.extend(["Lighter", "Knife", "Rope"])
    elif item_select.upper().strip() == "B" :
        inventory.extend(["First aid kit", "Knife", "Lighter" ])
    elif item_select.upper().strip() == 'C':
        inventory.extend(["Lighter", "First aid kit", "Rope"])
    else:
        print('\nPlease enter correct selection (A,B or C)!')

# ========================================================================
# MAIN GAME LOOP
# ========================================================================

while day <= 5 and health > 0 and searching_for_help_attempt < 6 :
    status_show()
    while True:
        action_select_daytime = input(blink("""
==================================================
                DAYTIMEE🌄🌄
==================================================

Select Action :
1. Search for food
2. Continue to search for help
3. Built a shelter
4. Eat
5. Use first aid kit

Enter your action (1-5) : """, 0.015))
        if action_select_daytime.strip() == "1" :
            search_food()
        elif action_select_daytime.strip() == "2" :
            blink("SEARCHING FOR HELP...", 0.035)
            blink('---> ENERGY -20 | SEARCH ATTEMPT +1', 0.035)
            energy -= 20
            searching_for_help_attempt += 1
        elif action_select_daytime.strip() == "3":
          if not shelter:
            built_shelter()
            shelter = True
          else:
            blink("ALREADY BUILT A SHELTER ! ⛺", 0.035)
        elif action_select_daytime.strip() == "4":
          if food_supply > 0 :
            if "Lighter" in inventory :
                blink('COOKING & EATING....🍽️🔥', 0.035)
                blink("""
---> HEALTH +40
---> ENERGY +15
                """, 0.035)
                health = min(100,health + 40)
                energy = min(150,energy + 15)
                food_supply -= 1
            else :
                blink("EATING RAW FOOD....🍽️", 0.035)
                blink("""
---> HEALTH +20
---> ENERGY +8
                """, 0.035)
                health = min(100,health  + 20)
                energy = min(150, energy + 8)
                food_supply -= 1
          else:
            blink('DOESN\tHAVE ENOUGH FOOD SUPPLY!', 0.035)


        elif action_select_daytime.strip() == "5" :
            if 'First aid kit' in inventory :
                blink("---> USING FIRST AID KIT.....➕➕➕", 0.035)
                blink("---> HEALTH RESTORED TO 100%", 0.035)
                health = 100
                inventory.remove('First aid kit')
            else :
                blink('DOESN\'tHAVE ANY FIRST AID KIT', 0.035)
        else :
            blink('Please enter correct selection (1-5)!', 0.035)
            continue
        break

    #Check if health is depleted or help can be found during daytime
    if health <= 0 or searching_for_help_attempt >= 6 :
      break


    # ----------------------------------NIGHTTIME----------------
    while True:
          action_select_night = input(blink("""
==================================================
                 NIGHTTIME🌙🌙
==================================================
Select Action :
1. Search for food
2. Continue to search for help
3. Sleep
4. Eat

Enter your action : """, 0.015))

          if action_select_night.strip() == "1" :
              blink('YOU ENCOUNTER A BEAR😱😱!!!', 0.035)
              fight_bear()
              energy -= 20
          elif action_select_night.strip() == "2" :
              blink("YOU ENCOUNTER A BEAR😱😱!!!", 0.035)
              fight_bear()
              blink("---> ENERGY -20", 0.035)
              searching_for_help_attempt += 1
              energy -= 20
          elif action_select_night.strip() == "3" :
              blink("--> SLEEPING...😴😴😴", 0.035)
              blink("""
---> HEALTH +30
---> ENERGY +30
              """, 0.035)
              health = min(100, health + 30)
              energy = min(150, energy + 30)
          elif action_select_night.strip() == "4":
            if food_supply > 0:
              if "Lighter" in inventory :
                blink('--->COOKING & EATING AT NIGHT....🍽️🔥',0.035)
                blink("""
---> HEALTH +40
---> ENERGY +15
---> ENERGY -20
                """, 0.035)
                health = min(100,health + 40)
                energy = min(150,energy + 15)
                food_supply -= 1
                energy -= 20
              else :
                  blink("EATING RAW FOOD AT NIGHT....🍽️🍽️🍽️", 0.035)
                  health = min(100,health  + 20)
                  energy = min(150, energy + 8)
                  blink("""
---> HEALTH +20
---> ENERGY +8
---> ENERGY -20

                  """, 0.035)
                  food_supply -= 1
                  energy -= 20
            elif action_select_night.strip() == "5" :
              if not 'First aid kit' in inventory :
                  blink('DOESN\tHAVE ENOUGH FOOD SUPPLY!', 0.035)
                  energy -= 20
          else :
              print ('Please enter correct selection (1-5)!')
              continue
          break
    #Event in day 3:THUNDERSTORM
    if day == 3 :
       blink("THERE ARE STORM INCOMING⛈️⛈️⛈️!!!!", 0.035)
       if shelter == True :
           blink("YOU HAVE SHELTERR!Save from the storm😮‍💨😮‍💨", 0.035)
       elif shelter == False :
           blink("""
YOU DON\'tHAVE A SHELTERRR😱😱😱

---> HEALTH -30
---> ENERGY -20
""", 0.035)
           health -= 30
           energy -= 20

    if energy < 0 :
      blink("⚠️ ENERGY DEPLETED!Exertion penalty: HEALTH -20",0.035)
      health -= 20

    day += 1
#=============================================================================
# FINAL OUTCOME
#=============================================================================
print("\n=============================================================================================")
#final part --- decided the player win or dead
if health <= 0 :
    blink("GAME OVERRR, YOU\'RE DEAD☠️☠️☠️", 0.035)
elif searching_for_help_attempt >= 6:
  blink("RESCUED! 🚁 A search and rescue helicopter spotted your signal! You escaped the island",0.035)
else :
    blink("CONGRATULATION!!! YOU SUCCESSFULLY OVERCAME THE CHALLENGES AND SURVIVED THE ISLAND🎉🎉🎉🎉", 0.035)
print("================================================================================================")
```
